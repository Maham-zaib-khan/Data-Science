##### Introduction to Functions
Definition:

A function is a block of code that performs a specific task.
Functions help in organizing code, reusing code, and improving readability.



```python
## syntax
def function_name(parameters):
    """Docstring""" #explains the functionality of function
    # Function body
    return expression
```

```python
def mode(a,b):
    ''' function to find modulus of numbers'''
    return a % b
print(mode(10,5))
print(mode(6,5))
```

```python
## function with multiple parameters

def add(a,b):
    return a+b

result=add(2,4)
print(result) 
    
```

```python
## factorial function
def factorial(n):
    fac=n 
    i=n
    while(i>1): 
        i=i-1 
        fac=fac*i
    return fac
```

```python
#factorial(4) 
factorial(8)
factorial(5)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name} Welcome To the paradise")

greet("Maham")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome To the paradise")

greet("maham") 
greet()
```

```python
### Variable Length Arguments
## Positional And Keywords arguments

def print_numbers(*numbers): # * means that argument contains single value i.e. list which is a positional argument. its order or postion matters
    for number in numbers:
        print(number,end=" ") 
```

```python
print_numbers(1,2,3,2)
```

```python
#use of positional argument 
intro=('Maham','2023-ag-9753', 4,'BSBI')
def welcome(* ag):
    for num in ag:
        print(num,end=" ")
    
    
welcome('Maham','2023-ag-9753',4,'BSBI')
welcome(20,'maham',30,'amna',65,'ayiza')
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Haseeb")
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number,end=" ")
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Maham") #arguments not in form of key-value pairs
```

```python
### Keywords Arguments

def print_details(**kwargs):  # parameters in form of keyword arguments
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Ayesha",age="20",country="Islamabad")
```

```python
print_details(degree="Bachler's", semester=4)
print_details(num1=2,num2=8,num3=9)
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f" Positional argument :{val}")
    
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Haseeb",name="Haseeb",age="21",country="Pakistan")
```

```python
#use of keyword argument 
def key(**values):
    for key,value in values.items():
        print(f"key= {key} value= {value}")
        
```

```python
key(name="Ayat",age=20,degree="BSBI") 
```
